# 📦 Python Tuples

A **tuple** is a built-in data structure used to store an ordered collection of items.

Tuples are similar to lists, but they are **immutable**, meaning their values cannot be changed after creation.

<b>Single-Element Tuples Need a Comma:</b> To create a tuple with only one item, you must include a trailing comma (e.g., my_tuple = (1,)). Without it, Python will just see the value inside the parentheses.

````python
# Tuple declarations
t = ()
print(type(t))
````

````python
# Mixed data types
t1 = (1,2,3,45.6,67)
t2 = ('Roja',1,2,45.5,56,1+2j,True)
print(type(t1))
print(type(t2))
````

````python
# Single-Element Tuples Need a Comma
t3 = (4)
print(type(t3))
t4 = (4,)
print(type(t4))
````

````python
# Count() returns number of occurances
t5 = (10,'Roja',45,35,10,23,10,10.45)
print(t5.count(10))

# Get the index of value
print(t5.index(10))
````

````python
# Indexing and slicing
print(t5[3])
print(t5[3:5])
print(t5[3:8])
print(t5[3:8:2])
print(t5[::-1])
````

````python
# Tuple can't be mutable - Immutable
print(t5[2])
t5[2] = 10
````

````python
t1*3  # Repeat 3 times
````

````python
# Tuple membership
print(1 in t1)
````

<h2>Unpacking tuples</h2>

````python
tup = (3,4,5)
a,b,c = tup
print(a)
print(b)
print(c)
````

````python
# number of variable should be equal of values
a,b = tup
print(a)
print(b)
````

````python
a,*d = tup
print(a)
print(d)
````

````python
a = 2
b = 4
print(a)
print(b)
a,b = b,a
print(a)
print(b)
````
