<h2 style="color:darkred">Number System in Python</h2>

<h2 style="color:#1766a8">1. Binary Number System</h2>
A number system with base or radix 2 is known as a binary number system. Only 0 and 1 are used to represent numbers in this system.

````python
bin (25)
````

````python
bin (30)
````

````python
bin (45)
````

````python
int (0b11001)     # To convert to decimal number int can be used
````

<h2 style="color:#1766a8">2. Octal Number System</h2>
Octal Number System is one in which the base value is 8. It uses 8 digits i.e. 0-7 for the creation of Octal Numbers. It is also a positional system i.e weight is assigned to each position.

````python
oct (25)
````

````python
oct (8)
````

````python
oct (7)
````

````python
oct (125)
````

````python
int (0o7)
````

````python
int(0o10)
````

<h2 style="color:#1766a8">3. Hexadecimal Number System</h2>
A number system with a base of 16 is called a hexadecimal number system. It is represented using numbers between 0 to 9 and the alphabets between A to F. As both numeric digits and alphabets are used in this system, it is also called an <b>alphanumeric system</b>.

````python
hex (34)
````

````python
hex (15)
````

````python
hex (14)
````

````python
hex (125)
````

````python
int (0x7d)
````
