<h2 style="color:darkred">Role of Underscore(_)</h2>

<div>Python automatically stores the value of the last expression in the interpreter to a particular variable called "_." </div>
<div>We can also assign these value to another variable if you want.</div>

```python
5 + 4
```

```python
_         # stores the result of the above expression
```

```python
_ + 6
```

```python
a = _     # assigning the value of _ to another variable
```

```python
a * 3
```
