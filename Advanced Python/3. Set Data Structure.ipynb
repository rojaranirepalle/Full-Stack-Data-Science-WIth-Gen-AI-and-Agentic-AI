# 🧩 Python Sets

A **set** is a built-in data structure used to store an unordered collection of unique items.

Sets are useful for:
- ✔ Removing duplicates  
- ✔ Membership testing  
- ✔ Mathematical set operations

````python
# Set declaration
s = {}
print(type(s))
s1 = set()
print(type(s1))
````

````python
s2 = {1,2,3,4.5,78,'Roja',1+3j}
print(type(s2))
print(s2)
````

````python
# Index won't possible
s3 = {1,20,3,4.5,78}
print(s3)
s3[1]
````

````python
# Add element for the set and takes only 1 argument
s3.add(5)
print(s3)
````

````python
# Duplicates are not allowed
s4 = {1,4,5,1,5}
s4
````

````python
# To clear the set use this function
s4 = {1,4,5,1,5,7,8}
s4.clear()
print(s4)
````

````python
# Copy the data from one set to other set
s4 = {1,4,5,1,5,7,8}
s5 = s4.copy()
print(s5)
````

````python
# Remove the element if it's exist other wise won't throws any error
s4.discard(4)
print(s4)
````

````python
# Pop with Index won't possible
s4 = {1,4,5,1,5,7,8}
s4.pop()       # Removes random value based on index
````

````python
s4.pop(3) # Won't work with index
````
