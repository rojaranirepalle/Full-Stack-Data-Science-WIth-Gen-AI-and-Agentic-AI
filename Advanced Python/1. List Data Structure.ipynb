# 📋 Python Lists

A **list** is a built-in, mutable data structure used to store an ordered collection of items.

Lists are one of the most commonly used data structures in Python because they are flexible and easy to use.

````python
# List can contain any type or mixed type of data
l2 = [1,2,3,4,['a','b','c'],1+3j,34.7,True]
````

````python
# Copy list data it's not a shallow copy ( both are different address )
l3 = l2.copy()
# Check the memory address
print(id(l3))
print(id(l2))
````

````python
import copy 
l4 = copy.deepcopy(l2)
print(id(l4))
print(id(l2))
````

````python
# Count the no of occurance of value
print(l1)
print(l1.count(34.5))
l1
````

````python
# Extend the elements as iterables at the last
l1.extend([10,20])
print(l1)
````

````python
# Difference between extend and append
l1.extend([10,20])  # Extend the elements as iterables at the last
l1.append([30,40])  # Append will append as object at the last
print(l1)
````

````python
# Get the index of first occurance
l1.index(34.5)
````

````python
# Get the index after specific mentioned index
print(l1.index(34.5,5))
# Check the 20 after index 15. If there is 20 not available after 15 it'll throws error
print(l1.index(20,15))
````

````python
# Insert element at the position 3
l1.insert(3,67)
````
