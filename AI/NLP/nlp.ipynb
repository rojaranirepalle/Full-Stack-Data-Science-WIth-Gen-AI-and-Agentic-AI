```python
AI = '''Artificial Intelligence refers to the intelligence of machines. This is in contrast to the natural intelligence of
humans and animals. With Artificial Intelligence, machines perform functions such as learning, planning, reasoning and
problem-solving. Most noteworthy, Artificial Intelligence is the simulation of human intelligence by machines.
It is probably the fastest-growing development in the World of technology and innovation. Furthermore, many experts believe
AI could solve major challenges and crisis situations.'''
```

```python
type(AI)
```

```python
from nltk.tokenize import word_tokenize
AI_tokens = word_tokenize(AI)
AI_tokens
```

```python
len(AI_tokens)
```

```python
from nltk.tokenize import sent_tokenize
AI_Sent = sent_tokenize(AI)
AI_Sent
```

```python
len(AI_Sent)
```

```python
from nltk.tokenize import blankline_tokenize
AI_blank = blankline_tokenize(AI)
AI_blank
```

```python
len(AI_blank)
```

```python
from nltk.tokenize import WhitespaceTokenizer
AI_whitespace = WhitespaceTokenizer().tokenize(AI)
AI_whitespace
```

```python
import numpy as np
from PIL import Image

# 1. Load an existing image into a NumPy array
img = Image.open("Documents/AIML/Projects/EILAP.png")
img_array = np.array(img)

# 2. Perform your image processing/manipulation on the array
# Example: Invert the pixel colors
processed_array = 255 - img_array

# 3. CRITICAL: Cast the array back to uint8 format
final_array = processed_array.astype(np.uint8)

# 4. Convert the processed NumPy array back to a PIL Image
output_image = Image.fromarray(final_array)

# 5. Save or display the resulting image
output_image.save("output.jpg")
output_image.show()
```
