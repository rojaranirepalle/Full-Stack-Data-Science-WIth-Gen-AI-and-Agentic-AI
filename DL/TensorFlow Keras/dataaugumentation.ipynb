```python
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.preprocessing.image import load_img, img_to_array
```

```python
data_gen = ImageDataGenerator(
    rotation_range=40,
    width_shift_range=0.2,
    height_shift_range=0.2,
    shear_range=0.2,
    zoom_range=0.2,
    horizontal_flip=True,
    fill_mode='constant'
)
img = load_img('../CNN/Happy or Sad/training/Happy/7.jpg')  # Load an image
```

```python
img
```

```python
x = img_to_array(img)  # Convert the image to a numpy array
x = x.reshape((1,) + x.shape)  # Reshape the array to (1, height, width, channels)

# Generate augmented images and display them
i = 0
for batch in data_gen.flow(x, batch_size=1,
                           save_to_dir='preview', save_prefix='aug', save_format='jpeg'):    
    i += 1
    if i > 30:  # Generate 30 augmented images
        break
```
