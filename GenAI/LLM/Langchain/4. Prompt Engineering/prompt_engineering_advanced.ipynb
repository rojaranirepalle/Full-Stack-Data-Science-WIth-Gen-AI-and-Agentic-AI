```python
import openai
import langchain
import os

from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate

from dotenv import load_dotenv,find_dotenv

_ = load_dotenv(find_dotenv())
os.environ["GOOGLE_API_KEY"] = os.getenv("GEMINI_API_KEY")
#openai.api_key = os.getenv("OPENAI_API_KEY")
```

```python
def get_completion(prompt, model="google_genai:gemini-3.6-flash", temperature=0.7):
    """
    Get a completion from the specified Gemini model using the provided prompt.
    
    Args:
        prompt (str): The input prompt for the model.
        model (str): The model to use for completion. Default is "google_genai:gemini-3.6-flash".
        temperature (float): Sampling temperature. Default is 0.7.
        max_tokens (int): Maximum number of tokens to generate, including this model's internal
            reasoning tokens. Default is 1024.
        
    Returns:
        str: The generated completion text.
    """
    # thinking_level="low" keeps this reasoning model from spending the whole max_tokens
    # budget on internal reasoning before it ever emits visible text.
    llm = init_chat_model(model, temperature=temperature, thinking_level="low")
    response = llm.invoke(prompt)

    content = response.content
    if isinstance(content, list):
        return "".join(block.get("text", "") for block in content if isinstance(block, dict))
    return content
```

```python
#Ask the model to solve a simple math problem step by step
# example for chain of thought reasoning, where the model is asked to explain its reasoning step by step
prompt = "Solve the following math problem step by step: What is 25 multiplied by 4?"
completion = get_completion(prompt)
print(completion)
```

```python
# self consistent reasoning, where the model is asked to explain its reasoning step by step and then provide a final answer

prompt = "Solve the following math problem step by step and provide a final answer: What is 20% of 150?"
completion = get_completion(prompt)
print(completion)
```

```python
# Self Consistency Prompting


# Generate multiple answers
prompt= """
What is 15% of 200?
give the answer using three different simple calculations.
"""
print(get_completion(prompt))


prompt= """
find the most consistant final answer:
1. 15% of 200 can be calculated by multiplying 200 by 0.15:
200 x 0.15 = 30

2. Another way to calculate 15% of 200 is by dividing 200 by 100 to get 1% and then multiplying by 15:
200 / 100 = 2
2 x 15 = 30

3. You can also find 15% of 200 by finding 10% and 5% separately and then adding them together:
10% of 200 = 200 x 0.10 = 20
5% of 200 = 200 x 0.05 = 10
20 + 10 = 30.
Final answer:
"""
print(get_completion(prompt))
```

```python
from langchain_core.prompts import PromptTemplate

# Define the template using Jinja2 double curly-brace syntax
template = "Give me a {{adjective}} {{topic}}."

# Initialize the PromptTemplate with the jinja2 format
prompt = PromptTemplate(
    template=template, 
    input_variables=["adjective", "topic"], 
    template_format="jinja2"
)

# Format the prompt with your desired variables
formatted_prompt = prompt.format(adjective="funny", topic="joke")
print(formatted_prompt)
# Output: Give me a funny joke.
```

```python
from langchain_core.prompts import PromptTemplate

# Define the template using f-string single curly-brace syntax
template = "Give me a {adjective} {topic}."

# Initialize the PromptTemplate with the f-string format
prompt = PromptTemplate(
    template=template,
    input_variables=["adjective", "topic"],
    template_format="f-string"
)

# Format the prompt with your desired variables
formatted_prompt = prompt.format(adjective="funny", topic="joke")
print(formatted_prompt)
# Output: Give me a funny joke.
```

```python
from langchain_core.prompts import StringPromptTemplate

class MyPrompt(StringPromptTemplate):
    template: str

    def format(self, **kwargs):
        return self.template.format(**kwargs)

prompt = MyPrompt(
    input_variables=["topic"],
    template="Explain {topic} in simple words.",
)
print(prompt.format(topic="Automobile Industry"))
```
