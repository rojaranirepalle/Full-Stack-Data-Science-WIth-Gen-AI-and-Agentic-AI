```python
import openai
import langchain
import os

from langchain.chat_models import init_chat_model
from langchain_core.prompts import ChatPromptTemplate

from dotenv import load_dotenv,find_dotenv

_ = load_dotenv(find_dotenv())
os.environ["GOOGLE_API_KEY"] = os.getenv("GEMINI_API_KEY")
#openai.api_key = os.getenv("OPENAI_API_KEY")
```

```python
#configure the LLM Model and chain
#llm = init_chat_model(model_name="gpt-3.5-turbo", temperature=0.7,max_tokens=200)

llm = init_chat_model("google_genai:gemini-3.6-flash", temperature=0.7)
```

```python
# Zero shot prompt

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("user", "Classify the following text as positive, negative, or neutral: {text}")
])

chain = prompt | llm
response = chain.invoke({"text": "The cs team was not too good"})
print(response)
```

```python
# Zero shot summarization prompt

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("user", "Analyze the following text and provide a summary of the key points: {text}")
])  

chain = prompt | llm
response = chain.invoke({"text": "The cs team was not too good. They struggled with meeting deadlines and their code quality was subpar. However, they showed improvement over time and eventually delivered a satisfactory product."})
print(response)
```

```python
#zero shot question answering prompt

prompt = ChatPromptTemplate.from_messages([
    ("system", "You are a helpful assistant."),
    ("user", "Answer the following question based on the provided text: {text}. Question: {question}")
])      

chain = prompt | llm
response = chain.invoke({
    "text": "The cs team was not too good. They struggled with meeting deadlines and their code quality was subpar. However, they showed improvement over time and eventually delivered a satisfactory product.",
    "question": "What were the main challenges faced by the cs team and how did they overcome them?"
})
print(response)
```

```python
# Few-shot prompt

from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    input_variables=["text"],
    template="Classify the following text as positive,Negative or Neutral {text}. Here are some examples: \n\nText: 'The cs team was not too good.'\nSentiment: Negative\n\nText: 'The cs team was excellent and delivered on time.'\nSentiment: Positive\n\nText: 'The cs team was average, but they showed improvement.'\nSentiment: Neutral\n\nNow classify the following text:\n\nText: {text}\nSentiment:"
)       

chain = prompt | llm
response = chain.invoke({"text": "The cs team was not too good"})
print(response)
```
