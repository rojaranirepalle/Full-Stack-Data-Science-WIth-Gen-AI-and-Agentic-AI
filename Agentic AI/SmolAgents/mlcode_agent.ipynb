```python
import os
import litellm
import gradio as gr
from smolagents import CodeAgent, DuckDuckGoSearchTool, LiteLLMModel
import google.generativeai as genai
```

```python
litellm._turn_on_debug()
```

```python
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())  # loads GOOGLE_API_KEY from Agentic AI/.env
```

```python
genai.configure(api_key=os.getenv("GOOGLE_API_KEY"))
```

```python
model = LiteLLMModel(
    model_id= "gemini/gemini-3-flash-preview",
    api_key= os.getenv("GOOGLE_API_KEY")
)
```

```python
ml_code_agent = CodeAgent(
    tools = [DuckDuckGoSearchTool()],
    additional_authorized_imports=['pandas','numpy','sklearn','sklearn.*','json'],
    model = model
)
```

```python
# Function to Generate ML Code Step-by-Step

def generate_ml_code(prompt):
    full_prompt = """
    create a step-by-step plan to write python code for svm model using sklearn to
    classify the iris dataset.
    """
    try:
        response = ml_code_agent.run(full_prompt)
        return response
    except Exception as e:
        return f"Error: {str(e)}"
```

```python
iface = gr.Interface(
    fn = generate_ml_code,
    inputs ="text",
    outputs = "text",
    title = "ML Code generator",
    description = "Enter a prompt to generate required user info"
)
```

```python
iface.launch()
```
