```python
import os
import warnings
import pandas as pd
warnings.filterwarnings("ignore")
```

```python
from langchain_experimental.agents import create_pandas_dataframe_agent
from langchain_anthropic import ChatAnthropic

#from langchain.agents import initialize_agent
```

```python
# creating an agent
from dotenv import load_dotenv, find_dotenv

load_dotenv(find_dotenv())  # loads ANTHROPIC_API_KEY from Agentic AI/.env

llm = ChatAnthropic(model="claude-haiku-4-5")
```

```python
# reading the titanic dataset
titanic_df = pd.read_csv("../Dataset/Titanic.csv")
titanic_df.head()
```

```python
#llm = OpenAI()
```

```python
#from langchain_openai import ChatOpenAI
l#lm = ChatOpenAI(temperature=1, model = "gpt-5.4-mini")
```

```python
agent = create_pandas_dataframe_agent(
    llm,
    titanic_df,
    verbose=True,
    agent_type="openai-tools",
    allow_dangerous_code=True # This is critical fix
)
```

```python
agent.run("How many rows are there")
```
