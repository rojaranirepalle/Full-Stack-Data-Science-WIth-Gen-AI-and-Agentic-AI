### Linear Regression - Tuning

```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn import preprocessing
from sklearn.linear_model import LinearRegression,Lasso,Ridge
```

```python
car_data = pd.read_csv('data_set/car-mpg.csv')
car_data.head()
```

```python
car_data.info()
```

```python
car_data.describe()
```

### Data Cleaning

```python
car_data.isnull().sum()
```

```python
car_data
```

```python
car_data = car_data.drop(['car_name'], axis = 1)
car_data['origin'] = car_data['origin'].replace({1: 'america', 2: 'europe', 3: 'asia'})
car_data = pd.get_dummies(car_data,columns = ['origin'])
```

```python
car_data = car_data.replace('?', np.nan)
car_data.info()
```

```python
car_data['hp'] = pd.to_numeric(car_data['hp'], errors='coerce')
car_data.isnull().sum()
```

```python
car_data = car_data.apply(lambda x: x.fillna(x.median()), axis = 0)
car_data.isnull().sum()
```

### Model Building

```python
x = car_data.drop(['mpg'],axis = 1)
y = car_data[['mpg']]
```

```python
#Scaling the data

X_s = preprocessing.scale(x)
X_s = pd.DataFrame(X_s, columns = x.columns) #converting scaled data into dataframe

y_s = preprocessing.scale(y)
y_s = pd.DataFrame(y_s, columns = y.columns) #ideally train, test data should be in columns
```

```python
x_train,x_test,y_train,y_test = train_test_split(X_s,y_s,train_size=0.2,random_state=42)
```

```python
x_train
```

```python
y_train
```

### 1. Linear Regression Model

```python
linear_model = LinearRegression()
linear_model.fit(x_train,y_train)
y_pred = linear_model.predict(x_test)
```

```python
y_pred
```

```python
for idx, col_name in enumerate(x_train.columns):
    print('The coefficient for {} is {}'.format(col_name, linear_model.coef_[0][idx]))
    
intercept = linear_model.intercept_[0]
print('The intercept is {}'.format(intercept))
```

### 2. Ridge Model

```python
#alpha factor here is lambda (penalty term) which helps to reduce the magnitude of coeff

ridge_model = Ridge(alpha = 0.3)
ridge_model.fit(x_train, y_train)

for idx, col_name in enumerate(x_train.columns):
    print('The coefficient for {} is {}'.format(col_name, ridge_model.coef_[idx]))
intercept_ridge = ridge_model.intercept_[0]
print('The intercept is {}'.format(intercept_ridge))
```

### 3. Lasso Model

```python
#alpha factor here is lambda (penalty term) which helps to reduce the magnitude of coeff

lasso_model = Lasso(alpha = 0.1)
lasso_model.fit(x_train, y_train)

for idx, col_name in enumerate(x_train.columns):
    print('The coefficient for {} is {}'.format(col_name, lasso_model.coef_[idx]))
intercept_lasso = lasso_model.intercept_[0]
print('The intercept is {}'.format(intercept_lasso))
```

### Score Comparision
Model score - r^2 or coeff of determinant

r^2 = 1-(RSS/TSS) = Regression error/TSS 


```python

# Linear Model
print('----Linear Regression----')
print(linear_model.score(x_train, y_train))
print(linear_model.score(x_test, y_test))

#Ridge
print('----Ridge Regression----')
print(ridge_model.score(x_train, y_train))
print(ridge_model.score(x_test, y_test))

print('----Lasso Regression----')
#Lasso
print(lasso_model.score(x_train, y_train))
print(lasso_model.score(x_test, y_test))
```
